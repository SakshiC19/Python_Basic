```python
# importing the required libraries

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# getting the built-in datasets of seaborn.

sns.get_dataset_names()
```

```python
# loading dataset

df= sns.load_dataset("exercise")
```

```python
df
```

```python
df.pulse.value_counts()
```

```python
#to remove Unnamed column 

df = df.drop(columns=["Unnamed: 0"])
```

```python
df.head()
```

```python
df.diet.unique()          # ['low fat', 'no fat']
```

```python
df.kind.unique()          # ['rest', 'walking', 'running']
```

```python
df.time.unique()          # ['1 min', '15 min', '30 min']
```

```python
df.kind.value_counts()    # 30 rows for each kind
```

```python
df.diet.value_counts()    # 45 rows for each diet
```

### 1 . Lineplot

```python
# average pulse as time increases

sns.lineplot(x='time', y='pulse', data=df, errorbar=None)
plt.title("Pulse over time")
plt.show()
```
