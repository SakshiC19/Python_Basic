### Filtering & Boolean Indexing

```python
Simple condition --->filter rows by matchig one condition
```

```python
df[df['age']>28]
```

```python
df
```

```python
#AND &

df[(df['age'] >25) & (df['city'] == 'Goa')]
```

```python
#OR /

df[(df['city'] == 'NY') | (df['name'] == 'David')]
```

```python
df[(df['city'] == 'NY') | (df['name'] == 'Bob')]
```

```python
#.isin() - Filter rows where a column matches a set of values

df[df['city'].isin(['NY','LA'])]
```

```python
#.query()  - Filter using readable string expression

df.query('age > 28 and city == "Goa" ')
```

```python
#.where() - Keep matching values , replace other with NAN 

df.where(df['age'] > 28)       #wherever condition not statisfied gives NAN
```

```python
#.mask()  -  it is reverse of where() i.e wherever condition true it gives NAN value

df.mask(df['age'] > 28)
```

### Handling Missing Data 

```python
df_na =pd.DataFrame({'a':[1,np.nan,3],'b':[np.nan,2,3]})
df_na
```

```python
#is_na gives output in boolean value form
#.isna() - Boolean mask of missing values

df_na.isna()
```

```python
#.notna() - Opposite of .isna()

df_na.notna()
```

```python
#df_na.isnull().sum() - Count of missing values per columns

df_na.isnull().sum()
```

```python
#.dropna() - Drop rows containing any NaN

df_na.dropna()            #it will only show rows without any NaN values
```

```python
#.dropna(axis=1)   - Drop columns containing any NaN

df_na.dropna(axis=1)
```

```python
#.fillna() - Fill missing values with constant

df_na.fillna(0)
```

```python
df_na
```

```python
#.ffill()  - Forward fill missing values

df_na.ffill()                     #at row 0 its 1.0 so forward will be row 1 instead NAN but at column b it cant be 0 because no forward   
```

```python
df_na  
```

```python
#.bfill()  - Backward fill missing values

df_na.bfill()
```

### Sorting & Ranking

```python
df=pd.DataFrame({
    'name':['Alice','Bob','Charlie','David'],
    'age':[25,30,35,40],
    'city':['NY','LA','SF','NY']
})
df
```

```python
df.sort_values(['city','age'],ascending=[True,False])
```

```python
#.sort_index()   ---> Sort row by the index

df.sort_index()
```

```python
#.rank()    ---> Rank values within a column

df['age'].rank
```

```python
#.nlargest()  - Top n rows by a column

df.nlargest(2,'age')     # two largest column from age column
```

```python
#.nsmallest()  - Bottom n rows by a column

df.nsmallest(3,'age')
```

### Aggregation and Statistics

```python
df=pd.DataFrame({
    'name':['Alice','Bob','Charlie','David'],
    'age':[25,30,35,40],
    'city':['NY','LA','SF','NY']
})
df
```

```python
df
```

```python
#.sum() / .mean() / .medium   ---> Basic Aggregations

df['age'].sum(), df['age'].mean(),df['age'].median()
```

```python
#.std() / .var()   --->  Spread of data

df['age'].std(),df['age'].var()
```

```python
# .min() / .max()  --->  Smallest and Largest value

df['age'].min(),df['age'].max()
```

```python
#.count()   ---> Number of not null value

df['age'].count()
```

```python
#.agg()  ---> Apply a multiple aggregation functions at once

df.agg({'age':['mean','max','min']})
```

```python
#.corr()  --> correlation matrix of numeric columns
#correlation is only applied to numerical column 
#correlation means comparsion between x-axis and y-axis

df.corr(numeric_only=True)
```

```python
#.mode()   ---> Most frequently occuring value

df['city'].mode()
```

### GroupBy Operations

```python
df
```

```python
g=df.groupby('city')        #groupby will create each and every value into a  one column 
```

```python
# .size   ---> Number of rows in each group

g.size()           #groupby +count
```

```python
#Group then aggregate a single column 

g['age'].mean()                   #avg age of people living in LA,NY,SF
```

```python
# .agg()   ---> On a group by --- Multiple aggregations per group

g.agg({'age':['mean','sum']})
```

```python
# .apply()    ---> On a groupby   --- custom function per group

g.apply(lambda x: x['age'].max() -x['age'].min())
```

#### Iterating over groups with a for loop

```python
g.size()                         # name and number of groups
```

```python
for name,group in g :
    print(name)
    print(group)
```

```python
# .get_group()   ---> Retrieve a specific group

g.get_group('NY')
```
